```python
import os
import numpy as np
import matplotlib.pyplot as plt

from PIL import Image

import tensorflow as tf
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, UpSampling2D
from tensorflow.keras.models import Model
```

```python
dataset_path = os.path.join(path, "mnist_png")
```

```python
def load_images(folder):

    images = []

    for digit in sorted(os.listdir(folder)):
        digit_folder = os.path.join(folder, digit)

        if os.path.isdir(digit_folder):

            for file in os.listdir(digit_folder):

                img = Image.open(os.path.join(digit_folder, file)).convert("L")
                img = img.resize((28,28))
                img = np.array(img, dtype=np.float32) / 255.0

                images.append(img)

    images = np.array(images).reshape(-1,28,28,1)

    return images
```

```python
train_images = load_images(os.path.join(dataset_path, "training"))
test_images = load_images(os.path.join(dataset_path, "testing"))

print(train_images.shape)
print(test_images.shape)
```
